## Implementing Simple and Composed Ops

```python
class Conv(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        # 3x3 conv, stride 1, padding 1: same resolution, new channel count
        self.net = nn.Sequential(nn.Conv2d(in_channels, out_channels, 3, 1, 1),
                                 nn.BatchNorm2d(out_channels), nn.GELU())

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class DownConv(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        # stride 2 halves the resolution: 28 -> 14 -> 7
        self.net = nn.Sequential(nn.Conv2d(in_channels, out_channels, 3, 2, 1),
                                 nn.BatchNorm2d(out_channels), nn.GELU())

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class UpConv(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        # transposed conv with kernel 4, stride 2, padding 1 doubles the resolution: 7 -> 14 -> 28
        self.net = nn.Sequential(nn.ConvTranspose2d(in_channels, out_channels, 4, 2, 1),
                                 nn.BatchNorm2d(out_channels), nn.GELU())

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class Flatten(nn.Module):
    def __init__(self):
        super().__init__()
        # average the whole 7x7 map down to 1x1
        self.net = nn.Sequential(nn.AvgPool2d(7), nn.GELU())

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class Unflatten(nn.Module):
    def __init__(self, in_channels: int):
        super().__init__()
        # kernel 7, stride 7 spreads the 1x1 vector back out to 7x7
        self.net = nn.Sequential(nn.ConvTranspose2d(in_channels, in_channels, 7, 7, 0),
                                 nn.BatchNorm2d(in_channels), nn.GELU())

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class ConvBlock(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        self.net = nn.Sequential(Conv(in_channels, out_channels), Conv(out_channels, out_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class DownBlock(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        self.net = nn.Sequential(DownConv(in_channels, out_channels), ConvBlock(out_channels, out_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class UpBlock(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        self.net = nn.Sequential(UpConv(in_channels, out_channels), ConvBlock(out_channels, out_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)
```

## Implementing Unconditional UNet

```python
class UnconditionalUNet(nn.Module):
    def __init__(
        self,
        in_channels: int,
        num_hiddens: int,
    ):
        super().__init__()
        D = num_hiddens
        self.in_block = ConvBlock(in_channels, D)    # D x 28 x 28
        self.down1 = DownBlock(D, D)                 # D x 14 x 14
        self.down2 = DownBlock(D, 2 * D)             # 2D x 7 x 7
        self.flatten = Flatten()                     # 2D x 1 x 1
        self.unflatten = Unflatten(2 * D)            # 2D x 7 x 7
        self.up1 = UpBlock(4 * D, D)                 # concat(2D, 2D) -> D x 14 x 14
        self.up2 = UpBlock(2 * D, D)                 # concat(D, D) -> D x 28 x 28
        self.out_block = ConvBlock(2 * D, D)         # concat(D, D) -> D x 28 x 28
        self.head = nn.Conv2d(D, in_channels, 3, 1, 1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        assert x.shape[-2:] == (28, 28), "Expect input shape to be (28, 28)."
        x0 = self.in_block(x)
        x1 = self.down1(x0)
        x2 = self.down2(x1)
        h = self.unflatten(self.flatten(x2))
        # skip connections: concatenate the matching encoder features along channels
        h = self.up1(torch.cat([x2, h], dim=1))
        h = self.up2(torch.cat([x1, h], dim=1))
        h = self.out_block(torch.cat([x0, h], dim=1))
        return self.head(h)
```

# Part 2.1: Implementing a Time-conditioned UNet

```python
class FCBlock(nn.Module):
    def __init__(self, in_channels: int, out_channels: int):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(in_channels, out_channels), nn.GELU(),
                                 nn.Linear(out_channels, out_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


class TimeConditionalUNet(UnconditionalUNet):
    def __init__(
        self,
        in_channels: int,
        num_classes: int,
        num_hiddens: int,
    ):
        super().__init__(in_channels, num_hiddens)
        D = num_hiddens
        # t is a single number, so F_in = 1
        self.fc1_t = FCBlock(1, 2 * D)
        self.fc2_t = FCBlock(1, D)

    def forward(
        self,
        x: torch.Tensor,
        t: torch.Tensor,
    ) -> torch.Tensor:
        """
        Args:
            x: (N, C, H, W) input tensor.
            t: (N,) normalized time tensor.

        Returns:
            (N, C, H, W) output tensor.
        """
        assert x.shape[-2:] == (28, 28), "Expect input shape to be (28, 28)."
        t = t.float().view(-1, 1)
        t1 = self.fc1_t(t)[:, :, None, None]   # (N, 2D, 1, 1)
        t2 = self.fc2_t(t)[:, :, None, None]   # (N, D, 1, 1)
        x0 = self.in_block(x)
        x1 = self.down1(x0)
        x2 = self.down2(x1)
        h = self.unflatten(self.flatten(x2)) * t1
        h = self.up1(torch.cat([x2, h], dim=1)) * t2
        h = self.up2(torch.cat([x1, h], dim=1))
        h = self.out_block(torch.cat([x0, h], dim=1))
        return self.head(h)
```

## Implementing the Forward and Reverse Process for Time-conditioned Denoising

```python
def time_fm_forward(
    unet: TimeConditionalUNet,
    x_1: torch.Tensor,
    num_ts: int,
) -> torch.Tensor:
    """Algorithm 1

    Args:
        unet: TimeConditionalUNet
        x_1: (N, C, H, W) input tensor.
        num_ts: int, number of timesteps.
    Returns:
        (,) loss.
    """
    unet.train()
    x_0 = torch.randn_like(x_1)
    t = torch.rand(x_1.shape[0], device=x_1.device)   # t ~ U[0, 1], one per image
    t_ = t.view(-1, 1, 1, 1)
    x_t = (1 - t_) * x_0 + t_ * x_1
    # the UNet predicts the flow x_1 - x_0, not the clean image
    return F.mse_loss(unet(x_t, t), x_1 - x_0)
```

```python
@torch.inference_mode()
def time_fm_sample(
    unet: TimeConditionalUNet,
    img_wh: tuple[int, int],
    num_ts: int,
    seed: int = 0,
    num_samples: int = 40,
) -> torch.Tensor:
    """Algorithm 2

    Args:
        unet: TimeConditionalUNet
        img_wh: (H, W) output image width and height.
        num_ts: int, number of timesteps.
        seed: int, random seed.
        num_samples: int, number of images to generate.

    Returns:
        (N, C, H, W) final sample.
    """
    unet.eval()
    dev = next(unet.parameters()).device
    # draw the starting noise on the CPU so the same seed gives the same digits on any device
    x = torch.randn((num_samples, 1, *img_wh), generator=torch.Generator().manual_seed(seed)).to(dev)
    for i in range(num_ts):
        t = torch.full((num_samples,), i / num_ts, device=dev)
        x = x + unet(x, t) / num_ts
    return x
```

```python
class TimeConditionalFM(nn.Module):
    def __init__(
        self,
        unet: TimeConditionalUNet,
        num_ts: int = 50,
        img_hw: tuple[int, int] = (28, 28),
    ):
        super().__init__()

        self.unet = unet
        self.num_ts = num_ts
        self.img_hw = img_hw


    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Args:
            x: (N, C, H, W) input tensor.

        Returns:
            (,) flow-matching loss.
        """
        return time_fm_forward(
            self.unet, x, self.num_ts
        )


    @torch.inference_mode()
    def sample(
        self,
        img_wh: tuple[int, int],
        seed: int = 0,
    ):
        return time_fm_sample(
            self.unet, img_wh, self.num_ts, seed
        )
```

# Part 2.2: Training the Time-conditioned UNet

```python
# Feel free to use code from part 1.2.1
# as they should be very similar
def train_flow(fm, name, sample, num_epochs=10, lr=1e-2, use_scheduler=True, batch_size=64,
               sample_epochs=(1, 5, 10)):
    """Train a flow-matching model; returns per-step losses and {epoch: samples}."""
    loader = make_loader(batch_size)
    optimizer = torch.optim.Adam(fm.parameters(), lr=lr)
    scheduler = (torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma=0.1 ** (1.0 / num_epochs))
                 if use_scheduler else None)
    losses, samples = [], {}
    start = time.perf_counter()
    for epoch in range(1, num_epochs + 1):
        for x, c in tqdm(loader, desc=f'{name} epoch {epoch}'):
            x, c = x.to(device), c.to(device)
            # only the class-conditioned model (defined in 2.4) takes the labels
            loss = fm(x, c) if hasattr(fm, 'p_uncond') else fm(x)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            losses.append(loss.item())
        if scheduler is not None:
            scheduler.step()   # once per epoch
        if epoch in sample_epochs:
            samples[epoch] = sample(fm)
            torch.save(fm.unet.state_dict(), CKPT / f'{name}_epoch{epoch}.pt')
    metrics[name] = {'final_loss': sum(losses[-100:]) / 100, 'train_seconds': time.perf_counter() - start,
                     'lr': lr, 'scheduler': use_scheduler}
    return losses, samples

torch.manual_seed(SEED)
time_fm = TimeConditionalFM(TimeConditionalUNet(1, 10, 64)).to(device)
time_losses, time_samples = train_flow(time_fm, 'time', lambda fm: fm.sample((28, 28), seed=0))
loss_plot(time_losses, 'b22', 'Time-conditioned UNet training loss')
```

# Part 2.4: Implementing a Class-conditioned UNet

```python
class ClassConditionalUNet(UnconditionalUNet):
    def __init__(
        self,
        in_channels: int,
        num_classes: int,
        num_hiddens: int,
    ):
        super().__init__(in_channels, num_hiddens)
        D = num_hiddens
        self.num_classes = num_classes
        self.fc1_t = FCBlock(1, 2 * D)
        self.fc2_t = FCBlock(1, D)
        self.fc1_c = FCBlock(num_classes, 2 * D)
        self.fc2_c = FCBlock(num_classes, D)

    def forward(
        self,
        x: torch.Tensor,
        c: torch.Tensor,
        t: torch.Tensor,
        mask: Optional[torch.Tensor] = None,
    ) -> torch.Tensor:
        """
        Args:
            x: (N, C, H, W) input tensor.
            c: (N,) int64 condition tensor.
            t: (N,) normalized time tensor.
            mask: (N,) mask tensor. If not None, mask out condition when mask == 0.

        Returns:
            (N, C, H, W) output tensor.
        """
        assert x.shape[-2:] == (28, 28), "Expect input shape to be (28, 28)."
        c = F.one_hot(c.long(), self.num_classes).float()
        if mask is not None:
            c = c * mask.float().view(-1, 1)   # a zero vector means "no class"
        t = t.float().view(-1, 1)
        c1, c2 = self.fc1_c(c)[:, :, None, None], self.fc2_c(c)[:, :, None, None]
        t1, t2 = self.fc1_t(t)[:, :, None, None], self.fc2_t(t)[:, :, None, None]
        x0 = self.in_block(x)
        x1 = self.down1(x0)
        x2 = self.down2(x1)
        # the class scales the features and the time shifts them
        h = c1 * self.unflatten(self.flatten(x2)) + t1
        h = c2 * self.up1(torch.cat([x2, h], dim=1)) + t2
        h = self.up2(torch.cat([x1, h], dim=1))
        h = self.out_block(torch.cat([x0, h], dim=1))
        return self.head(h)
```

```python
def class_fm_forward(
    unet: ClassConditionalUNet,
    x_1: torch.Tensor,
    c: torch.Tensor,
    p_uncond: float,
    num_ts: int,
) -> torch.Tensor:
    """Algorithm 3

    Args:
        unet: ClassConditionalUNet
        x_1: (N, C, H, W) input tensor.
        c: (N,) int64 condition tensor.
        p_uncond: float, probability of unconditioning the condition.
        num_ts: int, number of timesteps.

    Returns:
        (,) loss.
    """
    unet.train()
    x_0 = torch.randn_like(x_1)
    t = torch.rand(x_1.shape[0], device=x_1.device)
    t_ = t.view(-1, 1, 1, 1)
    x_t = (1 - t_) * x_0 + t_ * x_1
    # drop the class for p_uncond of the images, so the same network also learns the unconditional flow
    mask = (torch.rand(x_1.shape[0], device=x_1.device) >= p_uncond).float()
    return F.mse_loss(unet(x_t, c, t, mask), x_1 - x_0)
```

```python
@torch.inference_mode()
def class_fm_sample(
    unet: ClassConditionalUNet,
    c: torch.Tensor,
    img_wh: tuple[int, int],
    num_ts: int,
    guidance_scale: float = 5.0,
    seed: int = 0,
) -> torch.Tensor:
    """Algorithm 4

    Args:
        unet: ClassConditionalUNet
        c: (N,) int64 condition tensor. Only for class-conditional
        img_wh: (H, W) output image width and height.
        num_ts: int, number of timesteps.
        guidance_scale: float, CFG scale.
        seed: int, random seed.

    Returns:
        (N, C, H, W) final sample.
        (N, T_animation, C, H, W) caches.
    """
    unet.eval()
    dev = next(unet.parameters()).device
    c = c.to(dev)
    no_class = torch.zeros(len(c), device=dev)
    x = torch.randn((len(c), 1, *img_wh), generator=torch.Generator().manual_seed(seed)).to(dev)
    caches = [x]
    for i in range(num_ts):
        t = torch.full((len(c),), i / num_ts, device=dev)
        u_uncond = unet(x, c, t, mask=no_class)
        u_cond = unet(x, c, t)
        u = u_uncond + guidance_scale * (u_cond - u_uncond)   # classifier-free guidance
        x = x + u / num_ts
        if (i + 1) % 10 == 0:
            caches.append(x)   # every 10th step, for the animation
    return x, torch.stack(caches, dim=1)
```

```python
class ClassConditionalFM(nn.Module):
    def __init__(
        self,
        unet: ClassConditionalUNet,
        num_ts: int = 300,
        p_uncond: float = 0.1,
    ):
        super().__init__()
        self.unet = unet
        self.num_ts = num_ts
        self.p_uncond = p_uncond

    def forward(self, x: torch.Tensor, c: torch.Tensor) -> torch.Tensor:
        """
        Args:
            x: (N, C, H, W) input tensor.
            c: (N,) int64 condition tensor.

        Returns:
            (,) loss.
        """
        return class_fm_forward(
            self.unet, x, c, self.p_uncond, self.num_ts
        )

    @torch.inference_mode()
    def sample(
        self,
        c: torch.Tensor,
        img_wh: tuple[int, int],
        guidance_scale: float = 5.0,
        seed: int = 0,
    ):
        return class_fm_sample(
            self.unet, c, img_wh, self.num_ts, guidance_scale, seed
        )
```

# Part 2.5 Training the Class-conditioned UNet

```python
# Train a fresh class-conditioned D=64 UNet for 10 epochs, batch size 64,
# Adam initial lr=1e-2, class dropout probability .1, exponential gamma=.1**(1/10).
# Save checkpoints at epochs 1,5,10. Then train an independent no-scheduler
# comparison; choose/document a constant learning rate and compare results.
digits = torch.arange(10).repeat(4)   # four rows of 0-9
sample_digits = lambda fm: fm.sample(digits, (28, 28), guidance_scale=5.0, seed=0)

torch.manual_seed(SEED)
class_fm = ClassConditionalFM(ClassConditionalUNet(1, 10, 64)).to(device)
class_losses, class_samples = train_flow(class_fm, 'class', sample_digits)
loss_plot(class_losses, 'b25', 'Class-conditioned UNet training loss')

# Without the scheduler the learning rate never decays from 1e-2, so instead I use a constant
# 1e-3 (where the scheduler would end up): stable updates late in training, slower at the start.
NO_SCHEDULER_LR = 1e-3
torch.manual_seed(SEED)
flat_fm = ClassConditionalFM(ClassConditionalUNet(1, 10, 64)).to(device)
flat_losses, flat_samples = train_flow(flat_fm, 'class_no_scheduler', sample_digits,
                                       lr=NO_SCHEDULER_LR, use_scheduler=False)
loss_plot(flat_losses, 'b26_no_scheduler', f'Class-conditioned, no scheduler (constant lr {NO_SCHEDULER_LR:g})')
```
